# Hunger Box 

* It's a application used in the large scale companies where we can download and install in our Mobile to have handy to save our time by standing and taking the tokens to have food.
* We can easily select and puchase whatever item we need.

**Advantages**

* Time Saving
* Self Ording
* Advance Booking
* Prior Notifications by seeing the app whether we have items available or not
* Advance Searching of available food/items.

**Dis-Advantages**

* Payment Transactions may get failed due to server issues/Bank issues.
* More Waiting time to take tokens
