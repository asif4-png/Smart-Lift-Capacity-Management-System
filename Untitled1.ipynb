```python
class Lift:
    def __init__(self, max_capacity, max_people):
        self.max_capacity = max_capacity   # maximum weight in kg
        self.max_people = max_people       # maximum number of people
        self.current_weight = 0            # current total weight
        self.people_count = 0              # current number of people

    def enter(self, person_weight):
        """Add a person's weight to the lift if possible."""
        if self.people_count >= self.max_people:
            print("❌ Lift Capacity is allowed to have only 8 Person's. 9th Person cannot entered into the lift due to restricted!")
            return False

        if self.current_weight + person_weight <= self.max_capacity:
            self.current_weight += person_weight
            self.people_count += 1
            print(f"✅ Person entered with {person_weight} kg.")
            return True
        else:
            print("❌ Entered person is not having the space into the lift! (Weight limit exceeded)")
            return False

    def status(self):
        """Show current weight, remaining capacity, and people count."""
        left_capacity = self.max_capacity - self.current_weight
        left_people = self.max_people - self.people_count
        print(f"Current weight: {self.current_weight} kg")
        print(f"People in lift: {self.people_count}")
        print('-'*70)
        print(f"Remaining capacity: {left_capacity} kg")
        print(f"Remaining people slots: {left_people}")
        return left_capacity, left_people


if __name__ == "__main__":
    lift = Lift(max_capacity=630, max_people=8)

    while True:
        weight = input("Enter person's weight (or type 'stop' to exit): ")

        if weight.lower() == "stop":
            print("⏹️ Lift usage ended/Forcefully requested to STOP. Restart the Program again to become Operational.")
            break

        try:
            person_weight = float(weight)
        except ValueError:
            print("⚠️ Invalid input! Program terminated.")
            break   # terminate loop immediately

        # Process entry
        success = lift.enter(person_weight)
        lift.status()

        # Stop if limits are reached
        if not success:
            print("Entered Weight is too high with respect to remaining Capacity. No more heavy weight can be allowed.")
            break
        elif lift.people_count >= lift.max_people:
            print("❌ Lift Capacity is allowed to have only 8 Person's. 9th Person cannot be entered into the lift due to restrictions!")
            break
```
